## 3. Success metric

*One metric you can defend. What number means 'good'?*

### Primary Metric

The primary success metric for this ranking task will be **Precision@K**.

Precision@K measures how many of the top K pages recommended by the system are actually worthwhile review opportunities according to the target or proxy being evaluated.

This metric fits the business problem because an SEO specialist has limited time. The goal is not to recommend a fixed number of pages just to fill the review queue, but to place the most valuable opportunities near the top. If only a smaller number of pages are strong opportunities, the specialist's remaining time can be used on other work rather than reviewing weak recommendations.

Therefore, performance at the **top of the ranked queue** is more important for this use case than general accuracy across all webpages.

## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

A simple rule-based system can be used as a baseline. For example, a rule might prioritize pages with high impressions, low CTR, and older content. Such rules are easy to understand and implement, and they may work well for simple situations.

However, content refresh decisions often depend on multiple interacting factors rather than a single threshold. A page may have very high impressions but already perform extremely well, while another page with slightly lower impressions may provide a better refresh opportunity because of its engagement, freshness, position, or recent performance trends. Representing all of these interactions with fixed rules would require many manually designed conditions and exceptions.

Machine learning provides an opportunity to learn these relationships from historical data instead of relying only on manually defined rules. However, ML should only be adopted if it demonstrates a meaningful improvement over the rule-based baseline using the chosen evaluation metric. If a simple rule performs equally well, the simpler solution would be preferred.
